```python
from gtts import gTTS
from IPython.display import Audio

text_to_speech = gTTS('''Hello, this is a speech-to-text testing paragraph. Today is Wednesday, October seventh, twenty twenty-six, and I am testing how accurately Google Speech Recognition can understand my voice. I can speak slowly, quickly, loudly, or softly, and the system should convert my speech into written text without changing the meaning. My favorite programming languages are Python and JavaScript, and I am currently learning machine learning, data science, generative AI, and agentic AI. For example, I can say numbers like 10, 25.5, 1000, and 99.99. I can also test punctuation by saying comma, period, question mark, and exclamation mark. How accurately can the system recognize this sentence? Let's find out! This paragraph contains common words, technical terms, numbers, and punctuation so that I can properly test the speech recognition system.
''',lang='hi',tld='com')

text_to_speech.save('text_to_speech_gtts.wav')
sound_file = 'text_to_speech_gtts.wav'

Audio(sound_file, autoplay= False)
```

### Text to Speech conversion using pyttsx3

#### Changing the gender using pyttsx3

```python
import pyttsx3
from IPython.display import Audio

text = '''Hello, this is a speech-to-text testing paragraph. Today is Wednesday, October seventh, twenty twenty-six, and I am testing how accurately Google Speech Recognition can understand my voice. I can speak slowly, quickly, loudly, or softly, and the system should convert my speech into written text without changing the meaning. My favorite programming languages are Python and JavaScript, and I am currently learning machine learning, data science, generative AI, and agentic AI. For example, I can say numbers like 10, 25.5, 1000, and 99.99. I can also test punctuation by saying comma, period, question mark, and exclamation mark. How accurately can the system recognize this sentence? Let's find out! This paragraph contains common words, technical terms, numbers, and punctuation so that I can properly test the speech recognition system.
'''

audio = pyttsx3.init()
audio.setProperty('rate', 150)
audio.setProperty('volume', 0.8)

# change the voices
voice = audio.getProperty('voices')

# 0 for male and 1 for female
audio.setProperty('voice', voice[0].id)      # for male voice   
#audio.setProperty('voice', voice[1].id)       # for female voice

# text-to speech conversion
audio.say(text)

# save the audio file
audio.save_to_file(text, 'test_male_Voice.mp3')
#audio.save_to_file(text, 'test_female_Voice.mp3')

audio.runAndWait()
```
