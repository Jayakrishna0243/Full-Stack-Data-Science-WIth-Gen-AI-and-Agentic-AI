- what is INBUILD CLASS vs USER DEFINE CLASS 

### METHOD ( SPECIAL METHOD & USER DEFINE METHOD) 

```python
class computer:        
    def config(self):      # self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb')

com1 = computer()

com1.config()
```

```python
class computer:        
    def __init__(self):      #self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb')

com1 = computer()
com2 = computer()
com3 = computer()

print(id(com1))
print(id(com2))
```

```python
class computer:        
    def config(self):      # self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb')
com1 = computer()

class computer:        
    def __init__(self):      #self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb')
com1 = computer()
```

```python
class computer:   
    
    def config(self):      # self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb') 
        
    def __init__(self):      #self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb')
        
com1 = computer()
```

```python
class computer:   
    
    def __init__(self):      # self is the object which user passing & it is the keyword we understand later
        print('i5', '8gb') 
        
    def config(self):      #self is the object which user passing & it is the keyword we understand later
        print('i7', '16gb')
        
com1 = computer()
com2 = computer()
com1.config()
```
