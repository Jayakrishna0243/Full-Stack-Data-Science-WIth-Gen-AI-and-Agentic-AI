# Instance method, class method, static method

```python
class student:  # class name 
        
    school = 'nit'   # class varibale
    
    def __init__(self, m1, m2, m3) :  #special method with 4 arg
        self.m1 = m1 #inst
        self.m2 = m2 #inst
        self.m3 = m3 #inst

    def mean(self):  #avg is instnce method   
        return (self.m1+self.m2+self.m3)/3
    
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.mean())
print(s2.mean())

print(s1.m1)
print(s1.m2)
print(s2.m2)
```

# class method

```python
class student: 
        
    school = 'nit' 
    
    def __init__(self, m1, m2, m3) : 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):  # instance method 
        return (self.m1+self.m2+self.m3)/3
    
    def info(cls):  # class method 
        return cls.school 
        

s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg()) 
print(s2.avg())

print(s1.info()) 
print(s2.info())

#print(student.info())
```

```python
class student:        
    school = 'nit'     
    def __init__(self, m1, m2, m3) : 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
        
    def avg(self):  # instance method 
        return (self.m1+self.m2+self.m3)/3  
        
    @classmethod   
    def info(cls):  # class method 
        return cls.school  
        
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

s1.info()
```

# static method

```python
class student: 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod
    def getSchool(cls):
        return cls.school
    
    @staticmethod 
    def info():
        print('this is student class') 
        
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(s2.avg())

print(s1.getSchool())
print(s2.getSchool())

s1.info()
s2.info()
```
